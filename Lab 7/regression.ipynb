```python
import numpy as np
from sklearn.linear_model import LogisticRegression
X=np.array([[1],[2],[3],[4],[5],[6],[7],[8]])
y=np.array([0,0,0,0,1,1,1,1])
model=LogisticRegression()
model.fit(X,y)
hours=[[5.5]]
prediction=model.predict(hours)
if  prediction[0]==1:
     print("student will pass")
else:
    print("student will fail")
probability=model.predict_proba(hours)
print("probability of fail:",probability[0][0])
print("probability of pass:",probability[0][1])
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso
X=np.array([[1],[2],[3],[4],[5],[6],[7],[8]])
y=np.array([2,4,5,8,10,12,14,16])
model=Lasso(alpha=0.1)
model.fit(X,y)
y_pred=model.predict(X)
print("Coeffient:",model.coef_)
print("Intercept:",model.intercept_)
new_value=[[9]]
prediction=model.predict(new_value)
print("prediction for X=9:",prediction[0])
```

```python
import numpy as np
from sklearn.linear_model import Ridge
X=np.array([[1],[2],[3],[4],[5],[6],[7],[8]])
y=np.array([2,4,5,8,10,12,14,16])
model=Ridge(alpha=0.1)
model.fit(X,y)
print("Coeffient:",model.coef_)
print("Intercept:",model.intercept_)
new_value=[[9]]
prediction=model.predict(new_value)
print("prediction for X=9:",prediction[0])
```

```python
import numpy as np
from sklearn.linear_model import ElasticNet
X=np.array([[1],[2],[3],[4],[5],[6],[7],[8]])
y=np.array([2,4,5,8,10,12,14,16])
model=ElasticNet(alpha=0.1,l1_ratio=0.5)
model.fit(X,y)
print("Coeffient:",model.coef_)
print("Intercept:",model.intercept_)
new_value=[[9]]
prediction=model.predict(new_value)
print("prediction for X=9:",prediction[0])
```

```python
from sklearn.metrics import mean_squared_error
actual = [10, 20, 30, 40, 50]
predicted = [12, 18, 33, 38, 52]
mse = mean_squared_error(actual, predicted)
print("Actual Values:   ", actual)
print("Predicted Values:", predicted)
print("Mean Squared Error:", mse)
```

```python
from sklearn.metrics import mean_absolute_error
actual = [10, 20, 30, 40, 50]
predicted = [12, 18, 33, 38, 52]
mae = mean_absolute_error(actual, predicted)
print("Actual Values:   ", actual)
print("Predicted Values:", predicted)
print("Mean Absolute Error:", mae)
```
