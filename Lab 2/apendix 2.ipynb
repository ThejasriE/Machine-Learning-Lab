```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-np.pi,np.pi,100)
y=np.sin(x)
plt.plot(x,y)
plt.xlabel("Angel")
plt.ylabel("Sine Value")
plt.title("Sine Function")
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df1=sns.load_dataset("iris")
sns.distplot(df1['sepal_width'], kde=False)
plt.title("distribution plot for sepal width")
plt.show()
sns.distplot(df1['sepal_width'], hist=False)
plt.title("Density Plot for Sepal Width")
plt.show()
```

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
x=np.arange(1,6)
fig, a = plt.subplots(2,2)
a[0][0].plot(x, x*x)
a[0][0].set_title("x²")
a[1][1].plot(x, x*x*x)
a[1][1].set_title("x³")
plt.tight_layout()
plt.show()
```
